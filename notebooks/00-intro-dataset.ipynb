## Source

Dataset: [Wine Reviews](https://www.kaggle.com/datasets/zynicide/wine-reviews/data) — scraped from WineEnthusiast magazine by Zack Thoutt (Kaggle).

Three files are available in `data/raw/`:

| File | Format | Rows | Columns |
|---|---|---|---|
| `winemag-data-130k-v2.csv` | CSV | 129,971 | 13 |
| `winemag-data_first150k.csv` | CSV | 150,930 | 10 |
| `winemag-data-130k-v2.json` | JSON | 129,971 | 13 |

The **v2 CSV** (and its JSON mirror) is the primary file for this project — it is the most complete version, containing taster metadata not present in the 150k file.

```python
import pandas as pd
from pathlib import Path

RAW = Path("../data/raw")
```

## winemag-data-130k-v2.csv (primary)

129,971 wine reviews. Each row is one review written by a WineEnthusiast taster.

```python
df_v2 = pd.read_csv(RAW / "winemag-data-130k-v2.csv", index_col=0)
print(f"Shape: {df_v2.shape}")
print(f"\nDtypes:\n{df_v2.dtypes}")
print(f"\nMissing values:\n{df_v2.isnull().sum()}")
df_v2.describe(include="all")
```

### Column descriptions

| Column | Type | Description | Missing |
|---|---|---|---|
| `country` | categorical | Country where the wine was produced | 63 |
| `description` | text | Taster's written review | 0 |
| `designation` | categorical | Vineyard sub-label within the winery | 37,465 |
| `points` | int | Score assigned by the taster (80–100) | 0 |
| `price` | float | Bottle price in USD | 8,996 |
| `province` | categorical | Province or state of origin | 63 |
| `region_1` | categorical | Specific wine-growing region | 21,247 |
| `region_2` | categorical | More specific sub-region | 79,460 |
| `taster_name` | categorical | Name of the WineEnthusiast reviewer | 26,244 |
| `taster_twitter_handle` | categorical | Twitter handle of the reviewer | 31,213 |
| `title` | text | Full wine label (includes vintage year) | 0 |
| `variety` | categorical | Grape variety (e.g. Pinot Noir, Chardonnay) | 1 |
| `winery` | categorical | Name of the producing winery | 0 |

**Key stats:**
- `points`: mean 88.4, std 3.0, range 80–100
- `price`: mean $35.4, std $41.0, range $4–$3,300 (heavy right skew)
- `country`: 43 unique values — US dominates with 54,504 reviews
- `variety`: 707 unique grape varieties

## winemag-data_first150k.csv (older version)

```python
df_150k = pd.read_csv(RAW / "winemag-data_first150k.csv", index_col=0)
print(f"Shape: {df_150k.shape}")
print(f"\nDtypes:\n{df_150k.dtypes}")
print(f"\nMissing values:\n{df_150k.isnull().sum()}")
df_150k.describe(include="all")
```
