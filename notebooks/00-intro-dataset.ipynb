### v2 vs first150k comparison

Four checks to determine whether the 150k dataset can enrich the v2:

1. **Column diff** — which columns are shared vs exclusive to each file.
2. **Row overlap via `description`** — inner join on review text to count matching rows.
3. **Exclusive rows** — rows in 150k with no match in v2 (potential new data).
4. **Value mismatches** — for matched rows, do shared column values agree?

```python
# 1. Column diff
cols_v2 = set(df_v2.columns)
cols_150k = set(df_150k.columns)
shared_cols = cols_v2 & cols_150k

print("=== 1. Column diff ===")
print(f"Shared ({len(shared_cols)}):       {sorted(shared_cols)}")
print(f"Only in v2 ({len(cols_v2 - cols_150k)}):   {sorted(cols_v2 - cols_150k)}")
print(f"Only in 150k ({len(cols_150k - cols_v2)}): {sorted(cols_150k - cols_v2)}")

# 2. Row overlap via description
matched = df_v2[["description"]].merge(df_150k[["description"]], on="description", how="inner")

print("\n=== 2. Row overlap via description ===")
print(f"Rows in v2:           {len(df_v2):>7}")
print(f"Rows in 150k:         {len(df_150k):>7}")
print(f"Matched (inner join): {len(matched):>7}")
print(f"Match rate v2:        {len(matched) / len(df_v2) * 100:.1f}%")
print(f"Match rate 150k:      {len(matched) / len(df_150k) * 100:.1f}%")

# 3. Exclusive rows in 150k
n_exclusive = (~df_150k["description"].isin(df_v2["description"])).sum()
print("\n=== 3. Exclusive rows in 150k ===")
print(f"Rows with no match in v2: {n_exclusive:>7}")

# Duplicate descriptions (context for check 4)
print("\n=== Duplicate descriptions ===")
print(f"v2   duplicates: {df_v2['description'].duplicated().sum()}")
print(f"150k duplicates: {df_150k['description'].duplicated().sum()}")
```

```python
# 4. Value mismatches on shared columns for matched rows
compare_cols = sorted(shared_cols - {"description"})
merged = df_v2[list(shared_cols)].merge(
    df_150k[list(shared_cols)],
    on="description",
    how="inner",
    suffixes=("_v2", "_150k"),
)

print(f"=== 4. Value mismatches (shared cols, {len(merged)} matched rows) ===")
any_mismatch = False
for col in compare_cols:
    left, right = f"{col}_v2", f"{col}_150k"
    diff = (merged[left] != merged[right]) & merged[left].notna() & merged[right].notna()
    n = diff.sum()
    if n:
        any_mismatch = True
        print(f"\n  {col}: {n} mismatches")
        display(merged[[left, right]][diff].head(3))

if not any_mismatch:
    print("  No mismatches — all shared column values agree on matched rows")
```

**Conclusion:** the 150k dataset is not worth merging into v2.

- It adds **no new columns** — v2 already has a strict superset.
- Only **56.6% of v2 rows** match by description text, and the description key is unreliable (53k duplicate descriptions in 150k).
- Even "matched" rows disagree on key values (`country`, `points`, `price`, `winery`), indicating these are **two independent scrapes** with real data differences, not just formatting variants.
- The 82k exclusive rows in 150k come without `title`, `taster_name`, or `taster_twitter_handle`.

**Decision: use `winemag-data-130k-v2.csv` as the sole data source for all subsequent notebooks.**

## winemag-data-130k-v2.json

```python
with open(RAW / "winemag-data-130k-v2.json") as f:
    data_json = json.load(f)

print(f"Type: {type(data_json)}")
print(f"Records: {len(data_json)}")
print(f"Fields: {list(data_json[0].keys())}")
```

### CSV vs JSON equivalence check

Both v2 files claim to carry the same 129,971 records. We verify this by:

1. Aligning column order (JSON field order differs from CSV).
2. Resolving dtype mismatches before comparing (the JSON parser loads `points` as `object` instead of `int64`).
3. Using `DataFrame.equals()`, which treats `NaN == NaN` as `True` — essential given the many nulls in this dataset.

```python
df_json = pd.DataFrame(data_json)

# Align column order to match CSV
df_json = df_json[df_v2.columns]

# Report any dtype mismatches
dtype_mismatches = {
    col: (df_v2[col].dtype, df_json[col].dtype)
    for col in df_v2.columns
    if df_v2[col].dtype != df_json[col].dtype
}
print("Dtype mismatches:", dtype_mismatches)

# Cast JSON columns to match CSV dtypes
for col, (csv_dtype, _) in dtype_mismatches.items():
    df_json[col] = df_json[col].astype(csv_dtype)

# Compare (reset index so positional alignment is explicit)
are_identical = df_v2.reset_index(drop=True).equals(df_json.reset_index(drop=True))
print(f"\nFiles identical after dtype alignment: {are_identical}")
```

```python
rng = np.random.default_rng(42)
sample_idx = sorted(rng.choice(len(df_v2), size=5, replace=False))

csv_sample = df_v2.reset_index(drop=True).iloc[sample_idx]
json_sample = df_json.iloc[sample_idx]

print("=== CSV rows ===")
display(csv_sample)

print("=== JSON rows (same positions) ===")
display(json_sample)

print("=== Diff (empty means identical) ===")
display(csv_sample.reset_index(drop=True).compare(json_sample.reset_index(drop=True)))
```

**Result:** the CSV and JSON are perfect mirrors. The only difference was a dtype artifact — `points` loads as `object` from JSON vs `int64` from CSV — not a data discrepancy.

The CSV (`winemag-data-130k-v2.csv`) will be used as the single ingestion source for all subsequent notebooks.
